## Part 3 — Values and variables

- `int`: integer; `float`: decimal number; `str`: text; `bool`: `True` or `False`; `None`: missing value.
- `=` assigns a value; `type(x)` checks its type.
- `/` performs division, `//` performs floor division, and `%` returns the remainder.

```python
artist = "Dua Lipa"
ms_played = 183000
skipped = False
country = None

print(artist, ms_played, skipped, country)
print(type(artist), type(ms_played), type(skipped), type(country))
```

```python
seconds = ms_played / 1000
print(seconds)

minutes = 183 // 60
rest = 183 % 60
print(minutes, rest)
```

### Exercise 1 (Core) — Milliseconds to `m:ss`

For `ms = 225000`, calculate `minutes` and `seconds` as integers.

Expected output: `3 45`.

```python
ms = 225000

whole_seconds = ms // 1000
minutes = whole_seconds // 60
seconds = whole_seconds % 60

print(minutes, seconds)
assert (minutes, seconds) == (3, 45)
assert isinstance(minutes, int) and isinstance(seconds, int)
```

## Part 4 — Strings

Strings store text. Useful methods include `strip()`, `lower()`, `upper()`, `title()`, and `replace()`.

Use f-strings to include variables in text. Indexing starts at 0, and `split()` separates a string into parts.

```python
raw_artist = "   dUA lIPA  "
print(raw_artist.strip())
print(raw_artist.lower())
print(raw_artist.upper())
print(raw_artist.strip().title())
print(raw_artist.replace("U", "u"))
```

### Exercise 2 (Core) — Clean an artist name

Clean `messy = "  the WEEKND "` to get `The Weeknd`. Create `message` using an f-string.

Expected message: `Now playing: The Weeknd`.

```python
messy = "  the WEEKND "

clean = messy.strip().title()
message = f"Now playing: {clean}"

print(repr(clean))
print(message)
assert clean == "The Weeknd"
assert message == "Now playing: The Weeknd"
```

### Exercise 2b (Stretch) — Parse a duration string

Use `split()` and `int()` to convert `"3:45"` into total seconds.

Expected output: `225`.

```python
duration_text = "3:45"

minute_text, second_text = duration_text.split(":")
total = int(minute_text) * 60 + int(second_text)

print(total)
assert total == 225
```

## Part 5 — Lists

Lists (`[]`) store ordered values. Useful operations include `len()`, `sum()`, `max()`, `min()`, `sorted()`, and `append()`.

Use slicing such as `[-2:]` to access the last two items.

```python
minutes_per_day = [42, 15, 63, 0, 88, 120, 97]
print(len(minutes_per_day))
print(minutes_per_day[0])
print(minutes_per_day[-2:])
print(sum(minutes_per_day), max(minutes_per_day), min(minutes_per_day))
print(sorted(minutes_per_day, reverse=True))
print(88 in minutes_per_day)
minutes_per_day.append(30)
print(minutes_per_day)
```

### Exercise 3 (Core) — A week of listening

Calculate the total listening minutes, daily average (one decimal place), top three days, and the weekend share as a percentage.

Expected output: `464 66.3 [140, 110, 60] 53.9`.

```python
week = [35, 20, 48, 51, 60, 140, 110]

total = sum(week)
average = round(total / len(week), 1)
top3 = sorted(week, reverse=True)[:3]
weekend_share = round((sum(week[-2:]) / total) * 100, 1)

print(total, average, top3, weekend_share)
assert (total, average, top3, weekend_share) == (464, 66.3, [140, 110, 60], 53.9)
```

## Part 6 — Dictionaries

A dictionary stores `key: value` pairs. Use `d[key]` to access a value, or `d.get(key, default)` to provide a fallback when a key is missing.

Dictionaries can also contain nested dictionaries.

```python
play = {
    "event_id": "7a28c2bc7c9e",
    "user_id": "U0525",
    "track_id": "T0028",
    "ms_played": 226000,
    "skipped": False,
}
print(play["user_id"])
play["device"] = "desktop"
play["ms_played"] = 225000
print(play)
print(list(play.keys()))
```

```python
print(play.get("country"))
print(play.get("country", "unknown"))
```

```python
event = {
    "event_id": "f502f73a3acf",
    "event_type": "search",
    "user": {
        "id": "U0624",
        "plan": "premium",
        "device": {"type": "mobile", "os": "android"},
    },
    "payload": {"query": "queen", "results": 27},
}
print(event["user"]["plan"])
print(event["user"]["device"]["os"])
```

### Exercise 4 (Core) — Work with an event

From `event`, extract the search `query`, the device's `app_version` (use `"unknown"` if missing), and whether the user has a premium plan (`is_premium`).

Expected output: `queen unknown True`.

```python
query = event["payload"]["query"]
app_version = event["user"]["device"].get("app_version", "unknown")
is_premium = event["user"]["plan"] == "premium"

print(query, app_version, is_premium)
assert (query, app_version, is_premium) == ("queen", "unknown", True)
```

## Part 7 — A list of dictionaries is a table

A list of dictionaries can represent a small table: each dictionary is a row, and its keys are the columns. This is also a useful starting point for understanding Spark DataFrames.

```python
plays = [
    {"user_id": "U0525", "artist": "Dua Lipa", "seconds": 226, "skipped": False},
    {"user_id": "U0277", "artist": "Drake", "seconds": 6, "skipped": True},
    {"user_id": "U0458", "artist": "Coldplay", "seconds": 280, "skipped": False},
]
print(len(plays), "rows")
print(plays[1])
print(plays[1]["artist"])
```

### Exercise 5 (Core) — Add a row and summarise

Add a play by `U0554` for **Queen**, lasting `354` seconds and not skipped.

Without using loops, calculate the total seconds and the number of skipped plays across all four rows.

Expected output: `4 866 1`.

```python
plays.append({
    "user_id": "U0554",
    "artist": "Queen",
    "seconds": 354,
    "skipped": False,
})

total_seconds = (
    plays[0]["seconds"] + plays[1]["seconds"]
    + plays[2]["seconds"] + plays[3]["seconds"]
)
skip_count = (
    plays[0]["skipped"] + plays[1]["skipped"]
    + plays[2]["skipped"] + plays[3]["skipped"]
)

print(len(plays), total_seconds, skip_count)
assert (len(plays), total_seconds, skip_count) == (4, 866, 1)
```

### Exercise 5b (Stretch) — Sneak peek

Calculate `total_seconds` in one line so it works for any number of rows. Create an alphabetically sorted list of unique artist names called `artists`.

```python
total_seconds = sum(row["seconds"] for row in plays)
artists = sorted({row["artist"] for row in plays})

print(total_seconds, artists)
assert total_seconds == 866
assert artists == ["Coldplay", "Drake", "Dua Lipa", "Queen"]
```
